```python
from scipy import signal
from scipy.signal import firwin

print("low-pass filter for FM channel select")
print("sample rate: 62.5kHz | stopband attenuation: 60db | transition band: 3kHz (17kHz > 20kHz)\n---")

n_taps, beta = signal.kaiserord(60, 3000 / 31250)
print(f"number of taps: {n_taps} | beta: {beta}")

taps = firwin(
    numtaps=77,
    cutoff=17,
    fs=62500,
    pass_zero=True,
    window=('kaiser', beta)
    )
print(f"taps: {taps}")
```
