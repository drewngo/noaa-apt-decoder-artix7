```python
from scipy import signal
from scipy.signal import firwin, freqz

import numpy as np
import matplotlib.pyplot as plt

import pickle
```

```python
print("low-pass filter for FM channel select")
print("sample rate: 62.5kHz | stopband attenuation: 60db | transition band: 3kHz (17kHz > 20kHz)\n---")

n_taps, beta = signal.kaiserord(60, 3000 / 31250)
print(f"number of taps: {n_taps} | beta: {beta}")

taps = firwin(
    numtaps=77,
    cutoff=17000,
    fs=62500,
    pass_zero=True,
    window=('kaiser', beta)
    )

with open("channel_select_fir.pkl", "wb") as f:
    pickle.dump(taps, f)

w, h = freqz(b=taps, fs=62500)
plt.figure(figsize=(10, 5))
plt.plot(w, 20 * np.log10(abs(h)), 'b')
plt.axvline(x=17000, color="green", ls="--")
plt.text(x=17250, y=-40, s="Passband", color="green", rotation=90, va="center")
plt.axvline(x=20000, color="red", ls="--")
plt.text(x=20250, y=-40, s="Stopband", color="red", rotation=90, va="center")
plt.axhline(y=-60, color="purple", ls="--")
plt.text(x=10000, y=-55, s="Attenuation target", color="purple", va="center")
plt.title("Frequency Response")
plt.xlabel("Frequency (hz)")
plt.ylabel("Gain (dB)")
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 4))
plt.stem(taps, basefmt="C0-")
plt.title("Impulse Response")
plt.xlabel("Tap Index")
plt.ylabel("Amplitude")
plt.grid(True)
```
