```python
import filter_design_lib as fd
```

```python
print("low-pass filter for FM channel select")
print("sample rate: 62.4kHz | transition band: 3kHz (17kHz > 20kHz) | stopband attenuation: 60db\n---")

channel_select = fd.FIRSpec(
    sample_rate=62400,
    passband_edge=17000,
    stopband_edge=20000,
    attenuation_db=60
)

taps = fd.design_lowpass_fir(channel_select, "channel_select_fir.pkl")
print(f"number of taps: {len(taps)}")

fd.plot_frequency_response(taps, channel_select)
fd.plot_impulse_response(taps)
```

```python
print("anti-aliasing filter for decimation")
print("sample rate: 62.4kHz | transition band: 2kHz (2.4kHz > 4.4kHz) | stopband attenuation: 60db\n---")

anti_alias = fd.FIRSpec(
    sample_rate=62400,
    passband_edge=2400,
    stopband_edge=4400,
    attenuation_db=60
)

taps = fd.design_lowpass_fir(anti_alias, "anti_alias1_fir.pkl")
print(f"number of taps: {len(taps)}")

fd.plot_frequency_response(taps, anti_alias)
fd.plot_impulse_response(taps)
```
